# Décimo Nono Exercício

```python
import numpy as np

print("Cáculo de angulo entre duas arrays aleatórias \n")

#Gera valoes aleat´rios na array entre 1 até 10 com tamanho até 20.
lista1 = np.random.randint(1,11, size = 20)
lista2 = np.random.randint(1,11, size = 20)

print("Lista 1:", lista1)
print("Lista 2:", lista2)

# Calcula o produto escalar entre as duas arrays usando np.dot().
# O np.linalg.norm() calcula a norma (tamanho/comprimento) de cada array.
# A fórmula abaixo utiliza o produto escalar e as normas para encontrar o cosseno do ângulo entre as arrays.
cosseno = np.dot(lista1, lista2) / (np.linalg.norm(lista1) * np.linalg.norm(lista2))

# O np.dot() é utilizado para calcular o produto escalar, enquanto o np.linalg.norm() calcula o comprimento de cada vetor, sendo ambos necessários para aplicar a fórmula do cosseno.
angulo = np.arccos(cosseno)

#Mostra o resultado com a a formatação até o valor 4 decimal.
print(f"Cosseno do ângulo: {cosseno:.4f}")
print(f"Ângulo em radianos: {angulo:.4f}")
```

Usei o NumPy porque ele facilita a criação e manipulação de arrays e a realização de cálculos matemáticos. As funções (np.dot()), (np.linalg.norm()) e (np.arccos()) permitem aplicar diretamente a fórmula do ângulo entre dois vetores, assim, a lógica fica mais simples, organizada e eficiente do que realizar todos esses cálculos manualmente.

```python
from scipy.optimize import minimize_scalar # A importação no (minimize) permite escalar minimamente as funções.


print("Encontrando a raíz de uma função\n")

# Criando a função de X
def f(x):
    return -(x**2) + 4*x + 1

#Otimizando para a função negativa de X.
funcaoNegativa = lambda x: -f(x)

# Minimizando a escalação para encontrar o resultado
# A função (minimize_scalar) foi usado para minimizar funções.
resultado = minimize_scalar(funcaoNegativa)


print("O valor máximo atigindo é\n\n" , resultado)
```

Essa lógica deixa um método mais otimizado para encontrar os valores dentro de uma raíz.

# Sexto Exercício

```python
import numpy as np
from scipy import linalg # Outra importação para que o (linalg) para matrizes complexas.

print("Gerando uma matriz inversa com uma matriz aleatória")

# Gerando uma matriz de 5x5 com valores aleatórios.
matriz = np.random.randn(5, 5)

print("Matriz Original\n")
print(matriz)

# Cáculo da matriz inversa
try:
  #O (linalg) é feito para realizar operação de álbegra linear.
  #O (inv()) faz com que a matriz seja invertida.
    matrizInversa = linalg.inv(matriz)
    print("\nMatriz Inversa \n")
    print(matrizInversa)
# O (linalgError) vai exivir a falha do erro caso aconteça no cáculo.
except linalg.LinAlgError:
    print("A matriz não possui inversa (é singular).")
```

COm o SciPy O SciPy, através do módulo linalg, é utilizado para realizar operações de álgebra linear, nesse caso calculando a matriz inversa com linalg.inv().

```python
import numpy as np
from scipy.integrate import odeint # Importação para o cáculo diferencial ordinária da segunda ordem.

print("EDO de 2ª Ordem \n")


# Função para criar a expressão matemática no exercício.
def sistema(z, x):
    y, dy = z
    return [dy, -y]

z0 = [2,0]
x = np.linspace(0, 10, 100)

#Cáculo para a equação diferencial ordinária de segunda ordem.
solucao = odeint(sistema, z0, x)

print(f"O valor de y em x=10 é aproximadamente: {solucao[-1, 0]:.4f}")
```

```python
import pandas as pd

print("Encontrando a mediana com valores nulos em Preços")

#Criando o dicionário Dados
dados9 = {
    'Produto': ['Notebook', 'Smartphone', 'Monitor', 'Teclado', 'Mouse'],
    'Preço': [3500.00, None, 800.00, None, 50.00],
    'Quantidade': [4, 8, 5, 10, 15]
}

#Criando o Dataframe com os dados.
dataFrame = pd.DataFrame(dados9)

# Calculo da mediana da coluna.
#A função do (median()) serve para o cáculo de uma meiana do DataFrame.
mediaPreco = dataFrame['Preço'].median()

# Substituíção dos valores nulos pela mediana calculada.
#O comando (fillna()) serve para fazer substuição dos vaores nulos pela a coluna feita.
dataFrame['Preço'] = dataFrame['Preço'].fillna(mediaPreco)

print(f"\nMediana encontrada: {mediaPreco}")
print("\nDataFrame Corrigido:\n", dataFrame)
```

A função (median()) foi utilizada para calcular a mediana dos preços disponíveis, ignorando automaticamente os valores nulos.

Depois, (fillna(mediaPreco)) substitui os valores ausentes pela mediana calculada, evitando a perda dessas informações.
A variável (mediaPreco) armazena o valor da mediana, enquanto a coluna (Preço) recebe os valores corrigidos.

Nesss estilo , foi muito bom para fazer a mediana com valores nulos de forma mais ismples de escrever a lógica.

```python
import pandas as pd

print("Registros com valores de RA duplicados\n")

# DataFrame com dados de (RA) reptido.
dados = {
    'RA': [1, 5 , 5 , 7 , 8],
    'Nome': ['Ana', 'Bruno', 'Carlos', 'Daniel', 'Eduardo']
}
dataFrame = pd.DataFrame(dados)

print(dataFrame)

#Verificação de colunas duplicadas.
#A função do (diplicated()) serve para ver os dados duplicados no DataFrame.
linhasDuplicadas = dataFrame[dataFrame['RA'].duplicated()]

print(linhasDuplicadas)
```

```python
import pandas as pd

print("Tabela Dinâmica\n")

# DataFrame com vendas de produtos por categoria.
dados = {
    'Categoria': ['Eletrônicos', 'Periféricos', 'Eletrônicos', 'Periféricos', 'Eletrônicos'],
    'Quantidade': [11, 7 , 14 , 4 , 38]
}
dataFrame = pd.DataFrame(dados)

# Criando a tabela dinâmica para somar as quantidades por categoria.
tabelaDinamica = dataFrame.pivot_table(index='Categoria', values='Quantidade', aggfunc='sum')

print(tabelaDinamica)
```

A função (pivot_table()) foi utilizada para criar uma tabela dinâmica, permitindo resumir os dados de maneira organizada.

O parâmetro (index='Categoria') define as categorias como base da organização, enquanto values='Quantidade' indica os valores que serão analisados.

```python
import pandas as pd

print("DataFrames unificados através do ID\n")

# DataFrame - Informações cadastrais
dadosAlunos = pd.DataFrame({
    'ID': [1,2,3],
    'Nome': ['Ana', 'Bruno', 'Carlos']
})

# DataFrame - Notas escolares
dadosNotas = pd.DataFrame({
    'ID': [1,2,3],
    'Nota': [8.5, 7.0, 9.0]
})

# Mesclação dos dois DataFrames utilizando a coluna comum 'ID'
# O (merge()) faz a mesclagem de duas tabelas ou mais.
dataFrameConsolidado = pd.merge(dadosAlunos, dadosNotas, on='ID')

print(dataFrameConsolidado)
```

O parâmetro (on='ID') indica ao Pandas qual coluna deve ser utilizada como referência para realizar essa união.

```python
import pandas as pd

print("\nDataFrame corrigido com Interpolação Linear\n")

# DataFrame contendo uma sequência de temperatura com dias faltantes
dados10 = {
    'Dia': [12, 13 , 14 , 15 , 16],
    'Temperatura': [15.0, None, None, 18.0, 19.0]
}

dataFrame = pd.DataFrame(dados10)
print("DataFrame com dados de temperatura faltantes\n", dataFrame)

# Aplicamos a interpolação linear para preencher os valores nulos
# O comando (interpolate()) faz a interpolação a partir de um método escolhido.
dataFrame['Temperatura'] = dataFrame['Temperatura'].interpolate(method='linear')

print("Data Frame com a modificação\n", dataFrame)
```

A interpolação linear calcula uma progressão entre as temperaturas disponíveis, preenchendo os dias que não possuem informações.

Dessa forma, a lógica permite recuperar dados faltantes de maneira aproximada, mantendo a sequência de temperaturas mais completa e adequada para análises posteriores.

```python
import pandas as pd

print("Mediana salarial por Departamento")

# Dados de exemplo com funcionários, seus cargos e salários.
dadosEmpresa = {
    'Nome': ['Carlos', 'Aline', 'Sofia', 'Mateus', 'Gabriel', 'Juliana'],
    'Departamento': ['TI', 'TI', 'RH', 'Vendas', 'RH', 'TI'],
    'Salário': [5000.00, 6000.00, 3500.00, 4000.00, 4500.00, 5500.00]
}

dataFrame = pd.DataFrame(dadosEmpresa)

# Agrupamos os dados pelo (Departamento).
# Càculo da mediana da coluna (Salário).
mediana = dataFrame.groupby('Departamento')['Salário'].median().reset_index()

print(mediana)
```

A função (groupby()) foi utilizada para separar os funcionários de acordo com cada departamento.

O (reset_index()) reorganiza o resultado em formato de tabela, facilitando a visualização dos dados.

Dessa forma, a lógica utiliza o Pandas para agrupar os funcionários e calcular automaticamente a mediana salarial de cada departamento.

# Sexto Exercício



---



```python
import numpy as np
import matplotlib.pyplot as plt

# Definindo os intervalos de X adequados para cada função
x1 = np.linspace(0, 5, 100)       # Para a exponencial (e^x).
x2 = np.linspace(0.1, 10, 100)    # Para o logaritmo (não pode ser 0).

# Calculando os valores de Y
y1 = np.exp(x1)  # e^x (Exponencial) .
y2 = np.log(x2)  # log natural de x .

# Criando a figura e os dois subplots (1 linha, 2 colunas).
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

# Plot no primeiro gráfico (Exponencial).
ax1.plot(x1, y1, color='crimson', label='$y = e^x$')
ax1.set_title('Função Exponencial')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.grid(True)
ax1.legend()

# Plot no segundo gráfico (Logaritmo).
ax2.plot(x2, y2, color='royalblue', label='$y = \log(x)$')
ax2.set_title('Função Logarítmica')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.grid(True)
ax2.legend()

plt.tight_layout()
plt.show()
```

Com o Numpy , usei para gerar os valores de (X) e calcular as funções exponencial (e^x) e logarítmica log(x), criando os dados necessários para os gráficos.

O Matplotlib foi utilizado para criar uma figura com dois gráficos lado a lado, permitindo visualizar e comparar as duas funções.

Os comandos (set_title()), (set_xlabel()), (set_ylabel()), (grid()) e (legend()) foram utilizados para identificar e organizar cada gráfico.

```python
import numpy as np
import matplotlib.pyplot as plt

# Gerando pontos de 0 a 2 pi para fazer uma onda completa de seno.
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# Criando os limites superior e inferior do intervalo de confiança.
erro = 0.15
SuperiorY = y + erro
InferiorY = y - erro

# Criando a linha principal do Seno
plt.plot(x, y, color='purple', label='$y = \sin(x)$', linewidth=2)

# Preenchendo a área do intervalo de confiança (sombreado)
plt.fill_between(x, InferiorY, SuperiorY, color='purple', alpha=0.2, label='Intervalo de Confiança')

plt.title('Gráfico de Seno com Intervalo de Confiança Sombreado')
plt.xlabel('X (Radianos)')
plt.ylabel('Y')
plt.grid(True)
plt.legend()
plt.show()
```

A variável (erro) define uma margem de 0.15, utilizada para criar os limites superior e inferior ao redor da curva principal.

O Matplotlib, através de (plot()), foi utilizado para desenhar a função seno, enquanto (fill_between()) preenche a região entre os limites, representando visualmente o intervalo de confiança.

Os comandos de título, rótulos, grade e legenda foram adicionados para facilitar a interpretação das informações apresentadas no gráfico.

Dessa forma, a lógica combina NumPy para realizar os cálculos da função e dos limites e Matplotlib para representar a curva e destacar visualmente a margem de variação ao redor do seno.

```python
import numpy as np
import matplotlib.pyplot as plt

# Definindo as categorias e os grupos de dados.
categorias = ['Ano 1', 'Ano 2', 'Ano 3']
grupoA = np.array([2-4])
grupoB = np.array([2, 5, 6])
grupoC = np.array([4, 6, 7])

# Plotando as barras de forma empilhada.
plt.bar(categorias, grupoA, color='royalblue', label='Setor A')

# A segunda série (B) começa acima dos valores da série A.
plt.bar(categorias, grupoB, bottom=grupoA, color='orange', label='Setor B')

# A terceira série (C) começa acima da soma de A e B.
plt.bar(categorias, grupoC, bottom=grupoA + grupoB, color='forestgreen', label='Setor C')

plt.title('Distribuição de Orçamento por Setor (Empilhado)')
plt.ylabel('Milhares de Reais (R$)')
plt.legend()
plt.show()
```

Usei o NumPy para organizar os valores dos três grupos e realizar a soma necessária para determinar a posição de cada camada das barras empilhadas.

A função (plt.bar()) foi utilizada para criar as barras, enquanto o parâmetro (bottom) define a altura inicial de cada grupo, fazendo com que eles sejam posicionados um sobre o outro.

As expressões (grupoA + grupoB) permitem calcular onde o Setor (C) deve começar, mantendo o empilhamento correto dos valores.

As categorias, cores, título, rótulo do eixo e legenda foram adicionados para facilitar a identificação e interpretação de cada setor.

# Décimo Segundo Exercício



---



```python
import numpy as np
import matplotlib.pyplot as plt

# Fazendo os valores de X de -10 a 10 para uma curva ampla.
x = np.linspace(-10, 10, 200)
y = x ** 2  # Função quadrática f(x) = x²

# Desenho  da linha principal
plt.plot(x, y, color='purple', linewidth=2.5, label='$y = x^2$')

# Aplicando o ajuste de limites (Zoom) para o foco.
plt.xlim(-5, 5) # Limita o eixo X .
plt.ylim(0, 26) # Limita o eixo Y esteticamente para a parábola caber perfeitamente.

plt.title('Função Quadrática com Limites Ajustados')
plt.xlabel('Eixo X (Foco em -5 a 5)')
plt.ylabel('Eixo Y')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```

Usei o NumPy para gerar 200 pontos entre -10 e 10 e calcular os valores da função (quadrática y = x²), criando os dados necessários para formar a parábola.
O Matplotlib foi utilizado para desenhar a curva e representar visualmente o comportamento da função.

As funções (xlim()) e (ylim()) foram utilizadas para ajustar os limites dos eixos, criando um zoom na região de maior interesse do gráfico.

Os comandos de (título, rótulos, grade e legenda) foram adicionados para organizar a visualização e facilitar a interpretação da parábola.

```python
import numpy as np
import matplotlib.pyplot as plt

# Criando 100 pontos aleatórios para as três dimensões.
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

# Criando a figura e configurando os eixos em 3D.
figura = plt.figure()
eixos3D = figura.add_subplot(projection='3d')

# Fazendo a dispersão 3D.
grafico3D = eixos3D.scatter(x, y, z, c=z, cmap='coolwarm', s=50, edgecolor='black', alpha=0.8)

# Rótulos específicos para cada um dos três eixos.
eixos3D.set_title('Dispersão Tridimensional de Pontos Aleatórios')
eixos3D.set_xlabel('Eixo X')
eixos3D.set_ylabel('Eixo Y')
eixos3D.set_zlabel('Eixo Z')

plt.show()
```

O Matplotlib foi utilizado para criar uma figura com três eixos através do (projection='3d'), permitindo visualizar os dados em três dimensões.

A função (ax.scatter()) cria a dispersão 3D, enquanto (c=z) faz com que as cores dos pontos sejam determinadas pelos valores do eixo (Z).

# Décimo Quarto Exercício



---



```python
import numpy as np
import matplotlib.pyplot as plt

# Definindo 4 categorias e valores aleatórios entre 10 e 100 com tamanho igual a 4.
regioes = ['Sudeste', 'Nordeste', 'Sul', 'Centro-Oeste']
valores = np.random.randint(10, 101, size=4)

# Desenhando as barras horizontais
# O (barh) inverte a orientação padrão.
plt.barh(regioes, valores, color='indianred', edgecolor='black', height=0.6)

plt.title('Distribuição de Resultados por Região')
plt.xlabel('Pontuação Obtida')
plt.ylabel('Regiões')
plt.grid(True, axis='x', linestyle='--') # Linhas de grade apenas na vertical
plt.show()
```

A função (plt.barh()) do Matplotlib foi utilizada para criar um gráfico de barras horizontais, facilitando a comparação das pontuações entre as regiões.

O parâmetro (height) controla a espessura das barras, enquanto (edgecolor) destaca seus contornos e melhora a visualização.

A função (grid()) foi configurada com (axis='x') para adicionar linhas de referência apenas na direção vertical, facilitando a leitura das pontuações.

```python
import numpy as np
import matplotlib.pyplot as plt

# Gerando o intervalo de X contínuo de 1 a 100 com 400 pontos.
x = np.linspace(1, 100, 400)
# Calculamos os valores de Y
y = 1 / x

# Desenho da linha curva entre X e Y.
plt.plot(x, y, color='darkblue', linewidth=2, label='$y = 1/x$')

plt.title('Comportamento de Função Racional ($y = 1/x$)')
plt.xlabel('Valores de X (1 a 100)')
plt.ylabel('Valores de Y')
plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```

COm o Numpy , fiz a geração de 400 pontos entre os espeçamentos 1 a 100 e fazer o cáculo que se pede na atividade , sendo ( y =1/x).

A divisão (1 / x) permite observar como o valor de (Y) diminui conforme (X) aumenta.

O Matplotlib foi utilizado para representar essa relação através de um gráfico de linha, mostrando visualmente o comportamento decrescente da função.

Os comandos de (título, rótulos, grade e legenda) foram utilizados para organizar o gráfico e facilitar a interpretação dos valores.

# Décimo Sexto Exercício

```python
import numpy as np
import matplotlib.pyplot as plt

# Criando os dados
x = np.linspace(0, 10, 100)
y = 2 * x + 5

# 2. Desenhando o gráfico de linha.
plt.plot(x, y, color='blue', label='y = 2x + 5')

# Adicionando títulos aos eixos e ao gráfico.
plt.xlabel('Eixo X (Tempo)')
plt.ylabel('Eixo Y (Crescimento)')
plt.title('Gráfico de Linha com Rótulo de Texto')

# Adicionando o rótulo de texto exatamente no ponto (x = 5.)
plt.text(5, 15, ' Ponto Importante (x=5)', fontsize=10, color='red', fontweight='bold')

plt.grid(True, linestyle=':')
plt.legend()
plt.show()
```

Usei o NumPy para gerar 100 pontos entre 0 e 10 e calcular os valores da função linear (y = 2x + 5), criando os dados necessários para o gráfico.

O Matplotlib foi utilizado para desenhar a linha e organizar a visualização com título, rótulos dos eixos, grade e legenda.

A função (plt.text()) foi utilizada para adicionar um rótulo de texto em uma posição específica do gráfico, destacando o ponto importante definido em (x = 5).

Os parâmetros (fontsize, color e fontweight) permitem personalizar a aparência do texto para que ele fique mais visível.

A função (plt.savefig()) foi utilizada para exportar o gráfico diretamente para um arquivo PDF, enquanto (bbox_inches='tight') ajuda a evitar cortes nas informações do gráfico.

Depois, (plt.close()) fecha a figura e libera os recursos utilizados na memória após a exportação.

# Décimo Oitavo Exercício

```python
import numpy as np
import matplotlib.pyplot as plt

# Definindo dados de lucros/prejuízos.
categorias = ['Trimestre 1', 'Trimestre 2', 'Trimestre 3', 'Trimestre 4', 'Trimestre 5']
valores = np.array([2500, -1200, 3100, -800, 1500])

# Definindo cores dinâmicas.
coresParaBarras = ['green' if v >= 0 else 'red' for v in valores]

# 3. Desenhando o gráfico de barras.
plt.bar(categorias, valores, color=coresParaBarras, edgecolor='black', alpha=0.8)

# Adicionando uma linha de referência no (y=0) para deixar a divisão nítida.
plt.axhline(0, color='black', linewidth=1)

plt.title('Balanço Financeiro por Período')
plt.ylabel('Valor (R$)')
plt.grid(True, axis='y', linestyle='--', alpha=0.5)
plt.show()
```

Usei o NumPy para armazenar os valores de lucros e prejuízos em uma array, permitindo trabalhar com os resultados de cada trimestre de forma organizada.

A compreensão (['green' if v >= 0 else 'red' for v in valores]) foi utilizada para definir automaticamente a cor de cada barra, representando valores positivos em verde e negativos em vermelho.

O Matplotlib, através de (plt.bar()), foi utilizado para criar o gráfico de barras e representar visualmente os resultados financeiros de cada período.

A função (axhline(0)) adiciona uma linha de referência no valor zero, deixando clara a separação entre lucros acima de zero e prejuízos abaixo de zero.

# Décimo Nono Exercício

```python
import numpy as np
import matplotlib.pyplot as plt

# Criando os dados das duas séries da atividade.
x = np.linspace(0, 10, 100)
y1 = x ** 2     # Série A
y2 = x ** 2.5   # Série B

# Fazendo múltiplas chamadas de plot no mesmo desenho
plt.plot(x, y1, 'g-', label='Série A ($x^2$)')
plt.plot(x, y2, 'b:', label='Série B ($x^{2.5}$)')

# Adicionando a legenda automática
plt.legend(loc='best')

plt.title('Comparação de Crescimento de Séries')
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')
plt.grid(True, linestyle=':')
plt.show()
```

Usei o NumPy para gerar 100 pontos entre 0 e 10 e calcular duas funções diferentes, (x² e x²·⁵), criando as duas séries que serão comparadas.

A função (plt.plot()) foi utilizada duas vezes para desenhar as duas séries no mesmo gráfico, permitindo visualizar diretamente a diferença entre seus crescimentos.

Os estilos (g-) e (b:) diferenciam visualmente as curvas, utilizando linhas com padrões distintos para facilitar a identificação.

A função plt.legend() adiciona automaticamente uma legenda baseada nos nomes definidos em cada label, enquanto os demais comandos organizam o gráfico.

# Vigésimo Exercício

```python
import numpy as np
import matplotlib.pyplot as plt

# Criando os dados para a parábola.
x = np.linspace(0, 10, 100)
y = x ** 2

# Desenhando a curva.
plt.plot(x, y, color='purple', linewidth=2.5, label='$y = x^2$')

# 3. Definindo o espaçamento personalizado.
escalaX = np.arange(0, 11, 2)
escalandoY = np.arange(0, 101, 20)

# Aplicando os tiques personalizados
plt.xticks(escalaX)
plt.yticks(escalandoY)

# Ativando a grade, sendo desenhada exatamente nas posições feitas.
plt.grid(True, linestyle='--', color='gray', alpha=0.5)

plt.title('Parábola com Grade de Espaçamento Personalizado')
plt.xlabel('Valores de X')
plt.ylabel('Valores de Y')
plt.legend()
plt.show()
```

Usei o NumPy para gerar os pontos da função quadrática (y = x²) e também para criar as escalas personalizadas dos eixos X e Y.

As funções (np.arange()) foram utilizadas para definir os intervalos das marcações, permitindo controlar de quanto em quanto os valores aparecem nos eixos.

O Matplotlib foi utilizado para desenhar a parábola e aplicar essas marcações através de (xticks()) e (yticks()).

A função (grid()) foi adicionada para criar uma grade alinhada às marcações, facilitando a leitura e identificação dos valores no gráfico.
